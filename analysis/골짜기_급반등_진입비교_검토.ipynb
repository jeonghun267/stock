# 골짜기 급반등 진입 비교 검토

## TL;DR
전략2 직접매수는 신호만 늘리고 전부 -2% 손절되어 채택하지 않는다. 최종 저점 후 10초 내 +0.6% 급반등 7건은 비교한 모든 직접 진입 방식이 0건 포착했다.

## Context & Methods
2026-07-23~24 KST 09:00~09:20 1초 원본을 사용했다. 신호 생성에는 현재 시점까지의 자료만 사용했고, 사후 최저가와 손익 경로는 평가에만 사용했다. -2% 하드스톱, 왕복 수수료·세금 0.21%, 왕복 슬리피지 0.10%를 적용했다.

```python
comparison = [
    ('현행 골짜기', 1, 1, 1, 0, 7),
    ('전략2 직접매수', 12, 11, 12, 0, 7),
    ('먼저 온 신호', 13, 12, 13, 0, 7),
    ('FAST 10초', 7, 7, 7, 0, 7),
    ('전략2 READY+FAST', 0, 0, 0, 0, 7),
]
for name, signals, before_low, stops, captured, quick_events in comparison:
    print(f'{name}: 신호 {signals}, 최종저점 전 {before_low}, 손절 {stops}, 급반등 포착 {captured}/{quick_events}')
```

## Takeaways
전략2는 BUY가 아니라 관찰용 READY 후보로만 검토한다. 현행 FAST도 실전 승격 근거가 없으므로 S03은 주문0 상태에서 추가 거래일 표본을 모아야 한다. 원래 요약 JSON의 최초 `quick_v_capture_pct`는 최종 저점 이후 여부를 보지 않은 과대 집계이므로 사용하지 않고, 독립 검증의 0/7을 사용한다.
